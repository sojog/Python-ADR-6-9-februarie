```python
try:
    print(10 / 0)
    print(int("hello"))

except ZeroDivisionError as e:
    print("e:", type(e), e)
    print("__cause__", e.__cause__)
    print("__trackback__", e.__traceback__)
    print("__context__", e.__context__)
    print("Eroare de tipul ZeroDivisionError a fost gasita")

except ArithmeticError  as e:
    print("Eroare de tipul ArithmeticError a fost gasita")

except ValueError  as e:
    print("Eroare de tipul ValueError a fost gasita")

except Exception  as e:
    print("Eroare de tipul Exception a fost gasita")
```
