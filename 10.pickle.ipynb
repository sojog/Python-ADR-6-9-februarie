```python
import pickle
```

# Scriere in fisier

```python
file_handler = open("fisier.txt", "w") 
#...
file_handler.close()
```

```python
class Masina:
    def __init__(self, nume):
        self.nume = nume
    def __str__(self):
        return f"Masina mea cu numele {self.nume}"
```

```python
obiect_masina = Masina("Margareta")
```

```python
with open("fisier.pkl", "wb") as file_handler:
    pickle.dump(obiect_masina, file_handler)
```

```python

with open("fisier.pkl", "rb") as file_handler:
    obiect_citit = pickle.load(file_handler)
```

```python
obiect_citit.nume
```
